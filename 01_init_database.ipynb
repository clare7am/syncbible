```python
# 创建表：经句

cursor.execute("""
CREATE TABLE verse (
    id TEXT PRIMARY KEY,
    book_id INTEGER NOT NULL,
    chapter INTEGER NOT NULL,
    verse INTEGER NOT NULL,
    text_en TEXT NOT NULL,
    text_cn TEXT NOT NULL,
    FOREIGN KEY (book_id) REFERENCES book(id)
);
""")
conn.commit()
```

```python
# 创建表：分词

cursor.execute("""
CREATE TABLE words (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    verse_id TEXT NOT NULL,
    order_index INTEGER NOT NULL,
    word TEXT NOT NULL,
    type TEXT DEFAULT 'word',
    FOREIGN KEY (verse_id) REFERENCES verse(id)
);
""")
conn.commit()
```

```python
# 填充：英文分词

words = [
    "An", "account", "of", "the", "genealogy", "of",
    "Jesus", "the", "Messiah", ",", "the", "son", "of",
    "David", ",", "the", "son", "of", "Abraham", "."
]

for i, word in enumerate(words, start=1):
    word_type = 'punct' if word in ',.' else 'word'
    cursor.execute("""
        INSERT INTO words (verse_id, order_index, word, type)
        VALUES (?, ?, ?, ?)
    """, ("Mt.1.1", i, word, word_type))

conn.commit()
```
