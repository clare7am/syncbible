```python
import os
import requests
from bs4 import BeautifulSoup

# ========== 1. 拼接 URL ==========
def build_bible_gateway_url(book: str, chapter: str) -> str:
    book_encoded = book.replace(" ", "%20")
    return (
        "https://www.biblegateway.com/passage/"
        f"?search={book_encoded}%20{chapter}"
        "&version=NRSVCE"
    )


# ========== 2. 请求网页 HTML ==========
def fetch_passage_html(book: str, chapter: str) -> str:
    url = build_bible_gateway_url(book, chapter)
    response = requests.get(url)
    response.raise_for_status()
    return response.text


# ========== 3. 提取经文文本 ==========
def extract_passage_text(html: str) -> str:
    soup = BeautifulSoup(html, "html.parser")

    target_div = soup.find(
        "div",
        class_="version-NRSVCE result-text-style-normal text-html"
    )

    if target_div is None:
        return ""

    # 移除脚注 / 章节号 / 标题
    for tag in target_div.find_all(class_=["footnote", "footnotes", "chapternum"]):
        tag.decompose()

    for h3 in target_div.find_all("h3"):
        h3.decompose()

    # 给每个 <p> 后面补一个空格，防止经文粘在一起
    for p in target_div.find_all("p"):
        p.append(" ")

    # ✅ 关键：用节号拆行
    for versenum in target_div.find_all("sup", class_="versenum"):
        versenum.insert_before("\n")
        versenum.decompose()

    # ✅ 清理空行 & 首行空行
    lines = [
        line.strip()
        for line in target_div.get_text(strip=False).splitlines()
        if line.strip()
    ]

    return "\n".join(lines)


# ========== 4. 主程序 ==========
if __name__ == "__main__":
    # 用户输入
    book = input("请输入书卷名（如 John）：").strip()
    chapter = input("请输入章号（如 3）：").strip()
    short_name = input("请输入书卷简称（如 Jn）：").strip()

    # 获取经文
    html = fetch_passage_html(book, chapter)
    text = extract_passage_text(html)

    # 确保输出目录存在
    output_dir = "outputs/plaintext"
    os.makedirs(output_dir, exist_ok=True)

    # 拼接文件名
    filename = f"{short_name}_{chapter}_en.txt"
    filepath = os.path.join(output_dir, filename)

    # 写入文件
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(text)

    print(f"✅ 已成功生成文件：{filepath}")
```
