```python
# 输出所有章的最后一节节号为json

import requests
import json
import sqlite3
import re
import time

DB_PATH = "db/bible.db"
BASE_URL = "https://www.wanyouzhenyuan.cn/index.php?m=bible&version=sigao&template={book_id}&chapter={chapter}"

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}


def fetch_max_verse(book_id, chapter):
    """抓取某一章的最大节号"""
    url = BASE_URL.format(book_id=book_id, chapter=chapter)
    try:
        resp = requests.get(url, headers=HEADERS, timeout=15)
        resp.encoding = resp.apparent_encoding
        html = resp.text
    except Exception as e:
        print(f"  ✗ 请求失败 book_id={book_id}, chapter={chapter}: {e}")
        return None

    matches = re.findall(r'<span\s+value="(\d+)"', html)
    if not matches:
        return None
    return int(matches[-1])


def main():
    conn = sqlite3.connect(DB_PATH)
    conn.row_factory = sqlite3.Row
    cur = conn.cursor()

    # ✅ 严格按照 book_id 读取 abbr_en 和 max_chapter
    cur.execute("""
        SELECT id AS book_id, abbr_en, max_chapter
        FROM book
        WHERE id BETWEEN 1 AND 73
        ORDER BY id
    """)
    books = cur.fetchall()
    conn.close()

    result = []

    for row in books:
        book_id = row["book_id"]
        abbr_en = row["abbr_en"]
        max_chapter = row["max_chapter"]

        print(f"📖 {book_id}. {abbr_en}（共 {max_chapter} 章）")

        for chapter in range(1, max_chapter + 1):
            max_verse = fetch_max_verse(book_id, chapter)
            if max_verse is None:
                print(f"  ⚠ 未找到节号 book_id={book_id}, chapter={chapter}")
                continue

            result.append({
                "book_id": book_id,
                "abbr_en": abbr_en,
                "chapter": chapter,
                "max_verse": max_verse
            })

            print(f"  ✓ {abbr_en} {chapter}:{max_verse}")
            time.sleep(0.3)  # 防止被封

    # ✅ 输出 JSON
    with open("bible_max_verse.json", "w", encoding="utf-8") as f:
        json.dump(result, f, ensure_ascii=False, indent=2)

    print(f"\n✅ 完成，共收录 {len(result)} 条，已保存 bible_max_verse.json")


if __name__ == "__main__":
    main()
```

```python
import json
import os

TXT_DIR = "outputs/plaintext"
JSON_FILE = "bible_max_verse.json"


def main():
    # 加载 json
    with open(JSON_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    # 构建查找表: (abbr_en, chapter) -> max_verse
    verse_map = {}
    for item in data:
        key = (item["abbr_en"], item["chapter"])
        verse_map[key] = item["max_verse"]

    # 遍历 txt 文件
    for filename in os.listdir(TXT_DIR):
        if not filename.endswith("_en.txt"):
            continue

        # 解析文件名
        try:
            name = filename[:-len("_en.txt")]
            abbr_en, chapter = name.split("_")
            chapter = int(chapter)
        except ValueError:
            print(f"⚠ 文件名格式异常，跳过: {filename}")
            continue

        key = (abbr_en, chapter)
        if key not in verse_map:
            print(f"⚠ json 中未找到对应记录，跳过: {filename}")
            continue

        max_verse = verse_map[key]

        # 统计 txt 行数
        filepath = os.path.join(TXT_DIR, filename)
        with open(filepath, "r", encoding="utf-8") as f:
            line_count = sum(1 for _ in f)

        # 比对
        if line_count != max_verse:
            print(f"❌ 不一致: {filename}")
            print(f"   txt 行数   : {line_count}")
            print(f"   json max_verse: {max_verse}")
            print()

    print("✅ 比对完成")


if __name__ == "__main__":
    main()
```

```python
import json
import os
from openpyxl import Workbook

TXT_DIR = "outputs/plaintext"
JSON_FILE = "bible_max_verse.json"
OUTPUT_EXCEL = "verse_mismatch.xlsx"


def main():
    # 加载 json
    with open(JSON_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    # 构建查找表: (abbr_en, chapter) -> max_verse
    verse_map = {}
    for item in data:
        key = (item["abbr_en"], item["chapter"])
        verse_map[key] = item["max_verse"]

    # Excel 数据（仅不一致）
    wb = Workbook()
    ws = wb.active
    ws.title = "不一致记录"
    ws.append(("txt文件名", "txt行数", "json_max_verse"))

    mismatch_count = 0

    for filename in sorted(os.listdir(TXT_DIR)):
        if not filename.endswith("_en.txt"):
            continue

        # 解析文件名
        try:
            name = filename[:-len("_en.txt")]
            abbr_en, chapter_str = name.split("_")
            chapter = int(chapter_str)
        except ValueError:
            print(f"⚠ 文件名格式异常，跳过: {filename}")
            continue

        key = (abbr_en, chapter)
        if key not in verse_map:
            print(f"⚠ json 中未找到对应记录，跳过: {filename}")
            continue

        max_verse = verse_map[key]

        # 统计 txt 行数
        filepath = os.path.join(TXT_DIR, filename)
        with open(filepath, "r", encoding="utf-8") as f:
            line_count = sum(1 for _ in f)

        # 仅在不一致时记录
        if line_count != max_verse:
            ws.append((filename, line_count, max_verse))
            mismatch_count += 1
            print(f"❌ 不一致: {filename}  txt={line_count}  json={max_verse}")

    # 保存或提示
    if mismatch_count > 0:
        wb.save(OUTPUT_EXCEL)
        print(f"\n✅ 共发现 {mismatch_count} 条不一致记录，已导出至: {OUTPUT_EXCEL}")
    else:
        print("\n✅ 全部一致，未生成 Excel 文件")


if __name__ == "__main__":
    main()
```
