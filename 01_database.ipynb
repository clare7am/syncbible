```python
# 从txt批量导入

import sqlite3

DB_PATH = "db/bible.db"
TXT_FILE = "db/max_chapter.txt"

with open(TXT_FILE, "r", encoding="utf-8") as f:
    values = [line.strip() for line in f if line.strip()]

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

for idx, val in enumerate(values, start=1):
    cursor.execute(
        "UPDATE book SET max_chapter = ? WHERE id = ?",
        (val, idx)
    )

conn.commit()
cursor.close()
conn.close()

print(f"✅ 成功更新 {len(values)} 行的 max_chapter")
```

```python
import sqlite3

db_path = "db/bible.db"

input_id = input("请输入要查询的字符串：").strip()

conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# 查询 timestamps.id
cursor.execute("""
SELECT 1 FROM timestamps WHERE id = ? LIMIT 1;
""", (input_id,))
has_in_timestamps = cursor.fetchone() is not None

# 查询 tokens.align_id
cursor.execute("""
SELECT 1 FROM tokens WHERE align_id = ? LIMIT 1;
""", (input_id,))
has_in_tokens = cursor.fetchone() is not None

if has_in_timestamps and has_in_tokens:
    print("✅ 两个表里都存在该 ID")
elif not has_in_timestamps and not has_in_tokens:
    print("❌ 两个表里都没有该 ID")
else:
    if not has_in_timestamps:
        print("❌ timestamps 表中不存在该 ID")
    if not has_in_tokens:
        print("❌ tokens 表中不存在该 ID")

conn.close()
```

```python
import sqlite3

# 删除指定前缀的数据
prefix = "01_Gen_004_"

conn = sqlite3.connect("db/bible.db")
cur = conn.cursor()

# 先查询数量
cur.execute(
    "SELECT COUNT(*) FROM timestamps WHERE id LIKE ?",
    (f"{prefix}%",)
)
count_before = cur.fetchone()[0]

if count_before == 0:
    print(f"⚠️ 没有找到 id 以 '{prefix}' 开头的数据，未执行删除")
else:
    # 执行删除
    cur.execute(
        "DELETE FROM timestamps WHERE id LIKE ?",
        (f"{prefix}%",)
    )
    conn.commit()
    print(f"✅ 已删除 {count_before} 行（id 以 '{prefix}' 开头）")

conn.close()
```
