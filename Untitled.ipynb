```python
import requests
from bs4 import BeautifulSoup


# 拼接 URL

def build_bible_gateway_url(book: str, chapter: str) -> str:
    """
    根据用户手动输入的书卷名和章号，
    拼接 Bible Gateway 的 passage 页面 URL
    """
    book_encoded = book.replace(" ", "%20")
    return (
        "https://www.biblegateway.com/passage/"
        f"?search={book_encoded}%20{chapter}"
        "&version=NRSVCE"
    )


# 请求网页 HTML

def fetch_passage_html(book: str, chapter: str) -> str:
    """
    根据书卷名和章号请求页面，
    返回完整的 HTML 源码
    """
    url = build_bible_gateway_url(book, chapter)
    response = requests.get(url)
    response.raise_for_status()
    return response.text


# 3. 提取指定 class 内容

def extract_passage_text(html: str) -> str:
    
    # 提取内容
    soup = BeautifulSoup(html, "html.parser")

    target_div = soup.find(
        "div",
        class_="version-NRSVCE result-text-style-normal text-html"
    )

    if target_div is None:
        return ""
    
    # 移除脚注
    for footnote in target_div.find_all(class_=["footnote", "footnotes", "chapternum"]):
        footnote.decompose()

    # 移除标题
    for h3 in target_div.find_all("h3"):
        h3.decompose()
        

    # 一节一行
    for versenum in target_div.find_all("sup", class_="versenum"):
        # 在节号前插入换行
        versenum.insert_before("\n")
        # 删除节号本身
        versenum.decompose()
        

    return target_div.get_text(strip=False)
        

# 让我输入

if __name__ == "__main__":

    # 提示用户输入书卷名
    book = input("请输入书卷名：").strip()

    # 提示用户输入章号
    chapter = input("请输入章号：").strip()

    # 获取页面 HTML
    html = fetch_passage_html(book, chapter)

    # 提取经文内容
    text = extract_passage_text(html)

    # 输出结果
    print(text)
```
